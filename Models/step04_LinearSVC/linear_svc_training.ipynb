```python
import pandas as pd
import numpy as np
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.model_selection import train_test_split
import joblib
```

```python
# Load dataset
df = pd.read_csv("D:\Major Project\XGBoost\Dataset\Cleaned_GPS_Spoofing_Dataset.csv")

# Final selected features
selected_features = ['TOW', 'PD', 'RX', 'TCD', 'CP', 'DO', 'PRN', 'CN0', 'PC', 'EC']

X = df[selected_features]
y = df['Output']

# Load scaler
scaler = joblib.load("D:\Major Project\XGBoost\step02_preprocessing\scaler.joblib")
X_scaled = scaler.transform(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)
```

```python
#Compute Class Weights - (Handling severe imbalance)
classes = np.unique(y_train)
class_counts = np.bincount(y_train)
total = len(y_train)

class_weights = {cls: total / class_counts[cls] for cls in classes}
class_weights
```

```python
svm_model = LinearSVC(
    class_weight=class_weights,
    C=1.0,
    max_iter=5000,
    verbose=1
)

svm_model.fit(X_train, y_train)
```

```python
y_pred = svm_model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
joblib.dump(svm_model, "svm_model.joblib")
print("Model saved as svm_model.joblib")
```
