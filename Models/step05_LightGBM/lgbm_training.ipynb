```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import lightgbm as lgb
import joblib

# Load data
df = pd.read_csv("D:\Major Project\XGBoost\Dataset\Cleaned_GPS_Spoofing_Dataset.csv")

# Use same selected features
selected_features = ['TOW', 'PD', 'RX', 'TCD', 'CP', 'DO', 'PRN', 'CN0', 'PC', 'EC']

X = df[selected_features]
y = df['Output']

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)

# LightGBM model
lgb_model = lgb.LGBMClassifier(
    objective='multiclass',
    num_class=4,
    boosting_type='gbdt',
    n_estimators=500,
    learning_rate=0.05,
    max_depth=-1,
    subsample=0.8,
    colsample_bytree=0.8,
    class_weight='balanced'
)

# Train
lgb_model.fit(X_train, y_train)

# Predict
y_pred = lgb_model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))

# Save model
joblib.dump(lgb_model, "lgbm_model.joblib")
print("Model saved as lgbm_model.joblib")
```
